```python

import cv2
import matplotlib.pyplot as plt

# Get the actual uploaded filename
filename = list(uploaded.keys())[0]

# Load the image
image = cv2.imread(filename)

# Convert BGR to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Display the original image
plt.figure(figsize=(8, 6))
plt.imshow(image_rgb)
plt.title('Original Dog Image')
plt.axis('off')
plt.show()
```

```python

# Convert the original image into grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Display the grayscale image
plt.figure(figsize=(8, 6))
plt.imshow(gray_image, cmap='gray')
plt.title('Grayscale Dog Image')
plt.axis('off')
plt.show()
```

```python

# Resize the original image
resized_image = cv2.resize(image, (300, 300))

# Display the resized image
plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(resized_image, cv2.COLOR_BGR2RGB))
plt.title('Resized Dog Image (300 x 300)')
plt.axis('off')
plt.show()

# Display image dimensions
print("Original Image Dimensions:", image.shape)
print("Resized Image Dimensions:", resized_image.shape)
```

```python

# Get image dimensions
height, width = image.shape[:2]

# Crop the center portion of the image
start_x = width // 4
start_y = height // 4
end_x = 3 * width // 4
end_y = 3 * height // 4

cropped_image = image[start_y:end_y, start_x:end_x]

# Convert BGR to RGB
cropped_rgb = cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB)

# Display cropped image
plt.figure(figsize=(8, 6))
plt.imshow(cropped_rgb)
plt.title('Cropped Dog Image')
plt.axis('off')
plt.show()

print("Original Dimensions:", image.shape)
print("Cropped Dimensions:", cropped_image.shape)

```

```python


# Apply strong Gaussian Blur
blurred_image = cv2.GaussianBlur(image, (51, 51), 0)

# Convert BGR to RGB
blurred_rgb = cv2.cvtColor(
    blurred_image, cv2.COLOR_BGR2RGB
)

# Display strongly blurred image
plt.figure(figsize=(8, 6))
plt.imshow(blurred_rgb)
plt.title('Strongly Blurred Dog Image')
plt.axis('off')
plt.show()

print("Strong Gaussian Blur Applied Successfully!")

```

```python

# Convert original image to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Apply Canny Edge Detection
edges = cv2.Canny(gray_image, 100, 200)

# Display the edge-detected image
plt.figure(figsize=(8, 6))
plt.imshow(edges, cmap='gray')
plt.title('Dog Image - Canny Edge Detection')
plt.axis('off')
plt.show()

print("Edge Detection Completed Successfully!")
```

```python

# Convert original image to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Apply binary thresholding
ret, threshold_image = cv2.threshold(
    gray_image, 127, 255, cv2.THRESH_BINARY
)

# Display threshold image
plt.figure(figsize=(8, 6))
plt.imshow(threshold_image, cmap='gray')
plt.title('Dog Image - Binary Thresholding')
plt.axis('off')
plt.show()

print("Image Thresholding Completed Successfully!")
```

```python

# Get image dimensions
height, width = image.shape[:2]

# Calculate the center of the image
center = (width // 2, height // 2)

# Create rotation matrix (45 degrees)
rotation_matrix = cv2.getRotationMatrix2D(
    center, 45, 1.0
)

# Rotate the image
rotated_image = cv2.warpAffine(
    image, rotation_matrix, (width, height)
)

# Convert BGR to RGB
rotated_rgb = cv2.cvtColor(
    rotated_image, cv2.COLOR_BGR2RGB
)

# Display rotated image
plt.figure(figsize=(8, 6))
plt.imshow(rotated_rgb)
plt.title('Rotated Dog Image - 45 Degrees')
plt.axis('off')
plt.show()

print("Image Rotation Completed Successfully!")
```

```python

# Flip image horizontally
horizontal_flip = cv2.flip(image, 1)

# Flip image vertically
vertical_flip = cv2.flip(image, 0)

# Convert images from BGR to RGB
horizontal_rgb = cv2.cvtColor(horizontal_flip, cv2.COLOR_BGR2RGB)
vertical_rgb = cv2.cvtColor(vertical_flip, cv2.COLOR_BGR2RGB)

# Display both flipped images
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(horizontal_rgb)
plt.title('Horizontal Flip')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(vertical_rgb)
plt.title('Vertical Flip')
plt.axis('off')

plt.show()

print("Image Flipping Completed Successfully!")
```

```python

# Increase brightness and contrast
bright_image = cv2.convertScaleAbs(
    image, alpha=1.5, beta=50
)

# Convert BGR to RGB
bright_rgb = cv2.cvtColor(
    bright_image, cv2.COLOR_BGR2RGB
)

# Display original and adjusted images
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Dog Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(bright_rgb)
plt.title('Increased Brightness and Contrast')
plt.axis('off')

plt.show()

print("Brightness and Contrast Adjusted Successfully!")
```
